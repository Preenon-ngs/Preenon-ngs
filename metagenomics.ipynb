```python
#Step 1: Install Required Tools
# Install SRA Toolkit and dependencies
!apt-get update
!apt-get install -y sra-toolkit fastqc

# Check installation
!fastq-dump --version
!fasterq-dump --version
```

```python
#Step 2: Create Working Directories
import os

# Create folders
os.makedirs("sra_data", exist_ok=True)
os.makedirs("fastq_files", exist_ok=True)
os.makedirs("qc_reports", exist_ok=True)
print("Directories created successfully")
```

```python
#Step 3: Define SRA Accession Number
# Example SRA accession
sra_accession = "SRR12345678"
print("Selected SRA:", sra_accession)
```

```python
#Step 4: Download SRA File from NCBI
# Download SRA file
!prefetch $sra_accession --output-directory sra_data
```

```python
#Step 5: Verify Downloaded Files
!ls -lh sra_data
```

```python
#Step 6: Convert SRA to FASTQ
# Convert SRA to FASTQ
!fasterq-dump $sra_accession \
    --split-files \
    --outdir fastq_files \
    --threads 4
```

```python
#Step 7: Compress FASTQ Files
# Compress FASTQ files
!gzip fastq_files/*.fastq
```

```python
#Step 8: Check Generated FASTQ Files
!ls -lh fastq_files
```

```python
#Step 9: Perform Quality Control Using FastQC
# Run FastQC
!fastqc fastq_files/*.gz -o qc_reports
```

```python
#Step 10: View QC Reports
!ls qc_reports
```

```python
#Step 11: Download QC Reports to Local System
from google.colab import files
# Download FastQC HTML report
files.download("qc_reports/SRR12345678_1_fastqc.html")
```

```python
#Step 12: Generate Metadata from SRA
# Fetch metadata
!vdb-dump --info $sra_accession
```

```python
#Step 13: Retrieve Multiple Metagenomic Samples
# List of SRA accessions
sra_list = [
    "SRR12345678",
    "SRR12345679",
    "SRR12345680"
]

for sra in sra_list:
    print(f"\nProcessing {sra}")

    # Download
    !prefetch $sra --output-directory sra_data

    # Convert to FASTQ
    !fasterq-dump $sra \
        --split-files \
        --outdir fastq_files \
        --threads 4

    # Compress
    !gzip fastq_files/{sra}*.fastq
```

```python
#Step 15: Disk Space Cleanup (Important in Colab)
# Remove SRA cache files
!rm -rf sra_data
# Check remaining storage
!df -h
```

```python
#STEP 1 — Define Existing FASTQ Files
R1 = "fastq_files/SRR12345678_1.fastq.gz"
R2 = "fastq_files/SRR12345678_2.fastq.gz"
print(R1)
print(R2)
```

```python
#STEP 2 — Install Bowtie2
!apt-get update -qq
!apt-get install -y bowtie2 samtools wget
```

```python
#STEP 3 — Create Output Directory
import os
os.makedirs("host_removed", exist_ok=True)
```

```python
#STEP 4 — Download Correct Human Genome
!wget -O human_reference.fa.gz \
https://ftp.ncbi.nlm.nih.gov/genomes/all/GCF/000/001/405/GCF_000001405.40_GRCh38.p14/GCF_000001405.40_GRCh38.p14_genomic.fna.gz
```

```python
#STEP 5 — Extract Genome
!gunzip -f human_reference.fa.gz
```

```python
#Verify:
!ls -lh human_reference.fa
```

```python
#STEP 6 — Build Bowtie2 Index
!bowtie2-build human_reference.fa human_db
```

```python
#STEP 7 — Remove Host Reads
!bowtie2 \
-x human_db \
-1 $R1 \
-2 $R2 \
--very-sensitive \
--threads 4 \
--un-conc-gz host_removed/clean_reads.fastq.gz \
-S host_removed/mapped.sam
```

```python
#STEP 8 — Verify Output
!ls -lh host_removed
```

```python
#STEP 1 — Define Correct Clean Read Filenames
R1 = "host_removed/clean_reads.fastq.1.gz"
R2 = "host_removed/clean_reads.fastq.2.gz"
print(R1)
print(R2)
```

```python
#STEP 2 — Verify Files Exist
!ls -lh host_removed
```

```python
#STEP 3 — Remove Old Failed Directories
!rm -rf kraken_db
!rm -rf kraken_results
!rm -rf metaphlan_results
!rm -rf visualization
!rm -rf assembly
```

```python
#STEP 5 — FIX Kraken2 PATH
import os
os.environ["PATH"] += ":/content/kraken2"
!kraken2 --version
```

```python
#STEP 6 — Download Prebuilt MiniKraken Database (FAST & STABLE)
!wget -O minikraken.tgz \
https://genome-idx.s3.amazonaws.com/kraken/k2_standard_08gb_20250402.tar.gz
```

```python
#STEP 7 — Extract Kraken2 Database
!mkdir -p kraken_db
!tar -xvzf minikraken.tgz -C kraken_db
```

```python
#STEP 8 — Verify Database
!find kraken_db | head
```

```python
#STEP 9 — Run Kraken2
!kraken2 \
--db kraken_db \
--paired \
$R1 \
$R2 \
--threads 4 \
--report kraken_results/report.txt \
--output kraken_results/output.txt
```

```python
#STEP 10 — Verify Kraken2 Output
!ls -lh kraken_results
```
